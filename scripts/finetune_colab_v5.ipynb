```
# 2. Fontok tesztelése
from PIL import Image, ImageDraw, ImageFont
from pathlib import Path

FONT_DIR = Path("/content/fonts")
TEST_TEXT = "őűŐŰ öüóőúéáűí ÖÜÓŐÚÉÁŰÍ"

# Összes font betöltése és tesztelése
FONTS = []
failed_fonts = []

for font_path in sorted(FONT_DIR.glob("*.ttf")):
    try:
        # Teszteljük hogy támogatja-e a magyar karaktereket
        font = ImageFont.truetype(str(font_path), 24)
        test_img = Image.new("RGB", (100, 30), "white")
        test_draw = ImageDraw.Draw(test_img)
        test_draw.text((5, 5), "őű", fill="black", font=font)
        
        # Ellenőrizzük hogy nem üres-e (néhány font nem támogatja)
        pixels = list(test_img.getdata())
        has_content = any(p != (255, 255, 255) for p in pixels)
        
        if has_content:
            FONTS.append((font_path.stem, str(font_path)))
            print(f"✓ {font_path.stem}")
        else:
            failed_fonts.append(font_path.stem)
            print(f"✗ {font_path.stem} (nem támogatja az ő/ű karaktereket)")
    except Exception as e:
        failed_fonts.append(font_path.stem)
        print(f"✗ {font_path.stem}: {e}")

print(f"\n✓ Használható fontok: {len(FONTS)}")
if failed_fonts:
    print(f"✗ Kihagyott fontok: {len(failed_fonts)}")

# Vizuális teszt
print("\nVizuális teszt:")
cols = 2
rows = (len(FONTS) + cols - 1) // cols
cell_width = 450
cell_height = 40
test_img = Image.new("RGB", (cell_width * cols, cell_height * rows), "white")
draw = ImageDraw.Draw(test_img)

for idx, (name, path) in enumerate(FONTS):
    col = idx % cols
    row = idx // cols
    x = col * cell_width + 10
    y = row * cell_height + 8
    
    font = ImageFont.truetype(path, 18)
    short_name = name[:15] + ".." if len(name) > 17 else name
    draw.text((x, y), f"{short_name}: {TEST_TEXT}", fill="black", font=font)

display(test_img)
```

```
# 3. Adatgenerálás
import json
import random
from pathlib import Path
from PIL import Image, ImageDraw, ImageFont

# Bővített magyar szókincs
HUNGARIAN_WORDS = [
    # ő karakterek
    "őr", "őriz", "ők", "ősz", "ősi", "őszinte", "őrült", "őröl", "ő",
    "erő", "idő", "mező", "tető", "fő", "nő", "bő", "hő", "jő", "lő", "sző",
    "belső", "külső", "felső", "alsó", "utolsó", "első", "hátsó", "mellső",
    "költő", "festő", "vezető", "eladó", "vásárló", "igazgató", "szerző",
    "börtön", "könyv", "költség", "közel", "között", "mögött", "fölött",
    "Győr", "Szőlő", "Erdő", "Hősök", "Tőzsde",
    "dőlt", "dől", "bedől", "kidől", "eldől", "feldől", "ledől",
    "töröl", "pörög", "görög", "török", "örök", "rögös",
    
    # ű karakterek
    "űr", "űrlap", "űrhajó", "gyűrű", "tűz", "fűz", "nyűg", "bűn", "tű",
    "gyűjt", "gyűjtemény", "gyűlés", "gyűlölet", "gyűrű", "gyümölcs",
    "tűnik", "fűszer", "fűszeres", "tűző", "fűző", "tűzoltó",
    "hűtő", "hűvös", "hűség", "hűtlen", "hűsít", "hűsítő",
    "szürke", "szűk", "szűr", "szűz", "szűkít", "szűkül",
    "fülke", "küld", "kürt", "süt", "sűrű", "bűvös", "bűnös",
    "működik", "működés", "műszer", "művelet", "művész", "műhely",
    
    # Dokumentum szavak
    "Csatornadíj", "vízdíj", "díj", "díjszabás", "tandíj", "tagdíj",
    "tükörfúrógép", "árvíztűrő", "ÁRVÍZTŰRŐ", "TÜKÖRFÚRÓGÉP",
    "halványszürke", "sötétszürke", "világosszürke",
    "fizetendő", "követendő", "teljesítendő", "megoldandó", "elvégzendő",
    "összeg", "összesen", "összesítés", "összegzés", "összetétel",
    "kedvezmény", "átutalás", "számlaszám", "ügyintéző", "ügyfélazonosító",
    "adószám", "cégjegyzékszám", "azonosító", "rendszám", "törzsszám",
    "határidő", "lejárat", "esedékesség", "érvényesség", "keltezés",
]

SENTENCE_TEMPLATES = [
    "Fizetendő összeg: {amount} Ft",
    "Csatornadíj: {amount} Ft",
    "Vízdíj alapdíj: {amount} Ft",
    "Kedvezmény összege: {amount} Ft",
    "Késedelmi kamat: {amount} Ft",
    "Tartozás összege: {amount} Ft",
    "Előleg összege: {amount} Ft",
    "Adószám: {taxid}",
    "Cégjegyzékszám: {companyid}",
    "Fizetési határidő: {date}",
    "Lejárat: {date}",
    "Érvényesség: {date}",
    "Számla azonosító: {invoice}",
    "Referencia szám: {ref}",
    "Mérőóra állás: {meter}",
    "IBAN: HU{iban}",
    "Bankszámlaszám: {bankaccount}",
    "Tranzakció azonosító: TRX-{trx}",
    "Ügyfél azonosító: {customerid}",
    "Szerződés szám: {contractid}",
]

def random_format():
    return {
        "amount": f"{random.randint(1,999)} {random.randint(100,999):03d}",
        "taxid": f"{random.randint(10000000,99999999)}-{random.randint(1,2)}-{random.randint(10,99)}",
        "companyid": f"{random.randint(1,9):02d}-{random.randint(1,9):02d}-{random.randint(100000,999999)}",
        "date": f"2025.{random.randint(1,12):02d}.{random.randint(1,28):02d}",
        "invoice": f"INV-{random.randint(100000,999999)}",
        "ref": f"REF-2025-{random.randint(100000,999999):06d}-HU",
        "meter": f"{random.randint(0,99999):08d}",
        "iban": f"{random.randint(10,99)} {random.randint(1000,9999)} {random.randint(1000,9999)} {random.randint(1000,9999)} {random.randint(1000,9999)}",
        "bankaccount": f"{random.randint(10000000,99999999)}-{random.randint(10000000,99999999)}-{random.randint(10000000,99999999)}",
        "trx": f"2025-{random.randint(1000000,9999999)}",
        "customerid": f"CUS-{random.randint(100000,999999)}",
        "contractid": f"CON-2025-{random.randint(10000,99999)}",
    }

def generate_text():
    lines = []
    
    # Véletlenszerű szavak
    words = random.sample(HUNGARIAN_WORDS, random.randint(5, 10))
    lines.append(" ".join(words))
    
    # Mondatok
    for _ in range(random.randint(4, 7)):
        template = random.choice(SENTENCE_TEMPLATES)
        lines.append(template.format(**random_format()))
    
    # Kritikus tesztsorok
    lines.append("öüóőúéáűí - ÖÜÓŐÚÉÁŰÍ")
    lines.append("Árvíztűrő tükörfúrógép")
    
    return "\n".join(lines)

def render_text(text, font_path, width=850, font_size=24):
    font = ImageFont.truetype(font_path, font_size)
    lines = text.split("\n")
    line_height = int(font_size * 1.5)
    height = len(lines) * line_height + 80
    
    bg_colors = ["white", "#fafafa", "#f5f5f5", "#fffef0", "#f0f8ff", "#fff8f0", "#f8fff8"]
    text_colors = ["black", "#111111", "#1a1a1a", "#222222", "#0a0a0a"]
    
    img = Image.new("RGB", (width, height), random.choice(bg_colors))
    draw = ImageDraw.Draw(img)
    
    y = 40
    for line in lines:
        draw.text((40, y), line, fill=random.choice(text_colors), font=font)
        y += line_height
    return img

# Generálás
Path("training_data/images").mkdir(parents=True, exist_ok=True)
annotations = []

NUM_SAMPLES = 600  # Több minta
font_usage = {name: 0 for name, _ in FONTS}

print(f"Generálás: {NUM_SAMPLES} kép, {len(FONTS)} font...")
for i in range(NUM_SAMPLES):
    text = generate_text()
    font_name, font_path = random.choice(FONTS)
    font_usage[font_name] += 1
    font_size = random.choice([18, 20, 22, 24, 26, 28])
    
    img = render_text(text, font_path, font_size=font_size)
    img.save(f"training_data/images/{i:05d}.png")
    
    annotations.append({
        "image": f"{i:05d}.png",
        "text": text,
        "font": font_name,
    })
    
    if (i+1) % 100 == 0:
        print(f"  {i+1}/{NUM_SAMPLES}")

with open("training_data/annotations.jsonl", "w", encoding="utf-8") as f:
    for a in annotations:
        f.write(json.dumps(a, ensure_ascii=False) + "\n")

print(f"\n✓ Generálva: {NUM_SAMPLES} kép")
print(f"\nFont eloszlás:")
for name, count in sorted(font_usage.items(), key=lambda x: -x[1]):
    pct = count / NUM_SAMPLES * 100
    print(f"  {name}: {count} ({pct:.1f}%)")
```

```
# 5. Modell betöltése
import torch
from transformers import AutoProcessor, AutoModelForImageTextToText

MODEL_ID = "lightonai/LightOnOCR-2-1B-base"

print(f"Modell betöltése: {MODEL_ID}")
model = AutoModelForImageTextToText.from_pretrained(
    MODEL_ID,
    torch_dtype=torch.bfloat16,
    device_map="auto",
)
processor = AutoProcessor.from_pretrained(MODEL_ID)
print(f"✓ Modell betöltve")
```

```
# 6. LoRA
from peft import LoraConfig, get_peft_model

lora_config = LoraConfig(
    r=16,
    lora_alpha=32,
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj"],
    lora_dropout=0.05,
    bias="none",
)

model = get_peft_model(model, lora_config)
model.print_trainable_parameters()
```

```
# 7. Dataset
import json
from datasets import Dataset
from PIL import Image

def load_data():
    data = []
    with open("training_data/annotations.jsonl", encoding="utf-8") as f:
        for line in f:
            entry = json.loads(line)
            data.append({
                "image_path": f"training_data/images/{entry['image']}",
                "text": entry["text"]
            })
    return Dataset.from_list(data)

def process_example(example):
    image = Image.open(example["image_path"]).convert("RGB")
    text = example["text"]
    
    image_inputs = processor.image_processor(image, return_tensors="pt")
    text_inputs = processor.tokenizer(
        text,
        return_tensors="pt",
        padding="max_length",
        max_length=512,
        truncation=True,
    )
    
    return {
        "pixel_values": image_inputs["pixel_values"].squeeze(0),
        "input_ids": text_inputs["input_ids"].squeeze(0),
        "attention_mask": text_inputs["attention_mask"].squeeze(0),
        "labels": text_inputs["input_ids"].squeeze(0),
    }

dataset = load_data()
processed_dataset = dataset.map(process_example, remove_columns=dataset.column_names)
print(f"✓ Dataset: {len(processed_dataset)} példa")
```

```
# 8. Training
from transformers import TrainingArguments, Trainer

training_args = TrainingArguments(
    output_dir="./lighton-hun-lora",
    num_train_epochs=5,
    per_device_train_batch_size=2,
    gradient_accumulation_steps=8,
    learning_rate=5e-5,
    warmup_ratio=0.1,
    logging_steps=20,
    save_steps=100,
    save_total_limit=2,
    bf16=True,
    remove_unused_columns=False,
    dataloader_pin_memory=False,
    report_to="none",
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=processed_dataset,
)

print(f"Tanítás: {len(processed_dataset)} kép, {len(FONTS)} font, 5 epoch")
print("Várható idő: ~20-30 perc T4 GPU-n")
trainer.train()
print("\n✓ Tanítás kész!")
```

```
# 9. Mentés
model.save_pretrained("./lighton-hun-lora")
merged_model = model.merge_and_unload()
merged_model.save_pretrained("./lighton-hun-merged")
processor.save_pretrained("./lighton-hun-merged")
print("✓ Modell mentve")
```

```
# 10. Teszt
print("Tesztelés különböző fontokkal...\n")

test_indices = [0, 100, 200, 300, 400, 500]
for idx in test_indices:
    if idx >= len(annotations):
        continue
    test_img = Image.open(f"training_data/images/{idx:05d}.png")
    
    inputs = processor.image_processor(test_img, return_tensors="pt")
    inputs = {k: v.to(merged_model.device) for k, v in inputs.items()}
    inputs["input_ids"] = processor.tokenizer("", return_tensors="pt")["input_ids"].to(merged_model.device)
    
    with torch.no_grad():
        outputs = merged_model.generate(**inputs, max_new_tokens=500, do_sample=False)
    
    result = processor.tokenizer.decode(outputs[0], skip_special_tokens=True)
    
    print(f"=== #{idx} - {annotations[idx]['font']} ===")
    display(test_img)
    print(f"OCR:")
    print(result[:400])
    print()
```
